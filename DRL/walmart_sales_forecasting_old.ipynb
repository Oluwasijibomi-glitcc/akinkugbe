## Section 2 — Data Merging & Cleaning

We merge all three datasets into a single flat dataframe, then handle missing values:
- **MarkDown columns:** NaN means no promotion was active → fill with 0
- **CPI / Unemployment:** small number of missing values → fill with column median
- **Negative Weekly_Sales:** represent returns/corrections → remove (very few rows)

```python
# ── Convert date strings to datetime objects ──────────────────────────────────
sales['Date']    = pd.to_datetime(sales['Date'])
features['Date'] = pd.to_datetime(features['Date'])

# ── Merge sales + features on Store, Date, IsHoliday ─────────────────────────
# IsHoliday exists in both files — merging on it avoids duplicate columns
df = sales.merge(features, on=['Store', 'Date', 'IsHoliday'], how='left')

# ── Merge in store-level metadata (Type and Size) ─────────────────────────────
df = df.merge(stores, on='Store', how='left')

print(f'Merged dataframe: {df.shape[0]:,} rows × {df.shape[1]} columns')
display(df.head())
```

```python
# ── Inspect missing values before cleaning ────────────────────────────────────
missing = df.isnull().sum()
print('Missing values per column:')
print(missing[missing > 0])

# ── MarkDown cols: NaN = no promotion running that week → replace with 0 ──────
markdown_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5']
df[markdown_cols] = df[markdown_cols].fillna(0)

# ── CPI and Unemployment: fill remaining NaN with column median ───────────────
for col in ['CPI', 'Unemployment']:
    df[col].fillna(df[col].median(), inplace=True)

# ── Drop rows with negative Weekly_Sales (data corrections, < 0.1% of rows) ──
df = df[df['Weekly_Sales'] >= 0].reset_index(drop=True)

print(f'\nAfter cleaning → {df.shape[0]:,} rows')
print(f'Remaining missing values: {df.isnull().sum().sum()}')
```

## Section 4 — Feature Engineering

We create additional features to help the models learn temporal and promotional patterns:
- **Week / Month / Year** — extracted from the Date column to capture seasonality
- **TotalMarkDown** — sum of all 5 markdown columns as a single promotional spend signal
- **Type** — encoded as ordinal integer (A=0, B=1, C=2) to reflect store size ordering

```python
# ── Time-based features extracted from the Date column ───────────────────────
df['Week']  = df['Date'].dt.isocalendar().week.astype(int)  # week of year (1–52)
df['Month'] = df['Date'].dt.month                           # month (1–12)
df['Year']  = df['Date'].dt.year                            # year (2010–2012)

# ── Total promotional markdown spend across all 5 anonymous markdown types ────
df['TotalMarkDown'] = df[markdown_cols].sum(axis=1)

# ── Encode store Type as ordinal integer ──────────────────────────────────────
# A-type are the largest stores; B medium; C smallest
# Ordinal encoding preserves this ordering for the models
df['Type'] = df['Type'].map({'A': 0, 'B': 1, 'C': 2})

# ── Define the final feature set used in all models ──────────────────────────
FEATURES = [
    'Store', 'Dept',                               # store and department identifiers
    'IsHoliday',                                   # holiday flag
    'Temperature', 'Fuel_Price', 'CPI',            # economic environment
    'Unemployment',                                # labor market
    'Size', 'Type',                                # store characteristics
    'Week', 'Month', 'Year',                       # time features
    'TotalMarkDown'                                # total promotional spend
] + markdown_cols                                  # individual markdown channels

TARGET = 'Weekly_Sales'

print(f'Total features used in modeling: {len(FEATURES)}')
print('Features:', FEATURES)
```

## Section 5 — Train / Test Split

> **Assignment requirement:** Use the **last 30 weeks** of data as the test set.

We sort by date and find the 30th-to-last unique week as our cutoff. All models are trained on data before this date and evaluated on data from this date onward.

```python
# ── Sort the full dataset chronologically ─────────────────────────────────────
df = df.sort_values('Date').reset_index(drop=True)

# ── Identify the cutoff date: start of the last 30 unique weeks ───────────────
unique_dates = df['Date'].unique()
cutoff_date  = unique_dates[-30]  # 30th from the end

print(f'Test set starts : {cutoff_date}')
print(f'Test set ends   : {unique_dates[-1]}')

# ── Split into train and test ─────────────────────────────────────────────────
train = df[df['Date'] < cutoff_date]
test  = df[df['Date'] >= cutoff_date]

X_train = train[FEATURES]
y_train = train[TARGET]
X_test  = test[FEATURES]
y_test  = test[TARGET]

print(f'\nTrain rows: {X_train.shape[0]:,} | Test rows: {X_test.shape[0]:,}')
```

## Section 6 — T1: Inference — What Factors Drive Weekly Sales?

We use **OLS (Ordinary Least Squares) regression** via `statsmodels` for inference.  
OLS is the standard tool here because it provides:
- **Coefficient estimates** — direction and magnitude of each factor's effect
- **p-values** — statistical significance of each factor
- **Confidence intervals** — uncertainty around each estimate
- **R²** — overall explanatory power of the linear model

```python
# ── Ensure all feature columns are numeric (required by statsmodels) ──────────
# IsHoliday is boolean → convert to int (0/1)
X_train_ols = X_train.copy().astype(float)
X_test_ols  = X_test.copy().astype(float)

# ── Add intercept column (statsmodels does not add it automatically) ──────────
X_train_ols = sm.add_constant(X_train_ols)
X_test_ols  = sm.add_constant(X_test_ols)

# ── Fit OLS regression on the training set ────────────────────────────────────
ols_model = sm.OLS(y_train.astype(float), X_train_ols).fit()

# ── Print full regression summary ────────────────────────────────────────────
# Key things to look at: coef (direction), P>|t| (significance), R-squared
print(ols_model.summary())
```

```python
# ── Visualize coefficients — red bars = statistically significant (p < 0.05) ─
results_df = pd.DataFrame({
    'coef'  : ols_model.params,
    'pvalue': ols_model.pvalues
}).drop('const').sort_values('coef')

# Color bars by significance
colors = ['tomato' if p < 0.05 else 'lightgrey' for p in results_df['pvalue']]

plt.figure(figsize=(10, 7))
plt.barh(results_df.index, results_df['coef'], color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('OLS Regression Coefficients\n(Red = statistically significant at p < 0.05)',
          fontsize=13)
plt.xlabel('Coefficient Value')
plt.tight_layout()
plt.savefig('plot7_ols_coefficients.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Print only the significant features ──────────────────────────────────────
sig = results_df[results_df['pvalue'] < 0.05].sort_values('pvalue')
print(f'\n{len(sig)} statistically significant features (p < 0.05):')
print(sig[['coef', 'pvalue']].to_string())
```

## Section 7 — T2: Prediction — Traditional ML Models

We train and compare three models, from simplest to most complex:
1. **Linear Regression** — baseline, assumes linear relationships
2. **Random Forest** — ensemble of decision trees, handles non-linearity well
3. **XGBoost** — gradient boosting, typically top performer on tabular data

Evaluation metric: **MSE** (as required) and **RMSE** (for interpretability in $).

```python
# ── Shared helper function: fit, predict, and print metrics ───────────────────
def evaluate_model(name, model, X_tr, y_tr, X_te, y_te):
    """
    Trains a model on X_tr/y_tr, predicts on X_te/y_te,
    and returns the model, predictions, RMSE, and R².
    """
    model.fit(X_tr, y_tr)
    preds = model.predict(X_te)
    mse   = mean_squared_error(y_te, preds)
    rmse  = np.sqrt(mse)
    r2    = r2_score(y_te, preds)
    print(f'{name:25s} → RMSE: ${rmse:>10,.0f}  |  MSE: {mse:>15,.0f}  |  R²: {r2:.4f}')
    return model, preds, rmse, r2


# Dictionary to collect all results for the final comparison
results = {}

# ── Model 1: Linear Regression (baseline) ────────────────────────────────────
lr_model, lr_preds, lr_rmse, lr_r2 = evaluate_model(
    'Linear Regression',
    LinearRegression(),
    X_train, y_train, X_test, y_test
)
results['Linear Regression'] = {'RMSE': lr_rmse, 'R2': lr_r2}

# ── Model 2: Random Forest ────────────────────────────────────────────────────
# 100 trees; n_jobs=-1 uses all CPU cores to speed up training
rf_model, rf_preds, rf_rmse, rf_r2 = evaluate_model(
    'Random Forest',
    RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    X_train, y_train, X_test, y_test
)
results['Random Forest'] = {'RMSE': rf_rmse, 'R2': rf_r2}

# ── Model 3: XGBoost ──────────────────────────────────────────────────────────
# 300 shallow trees with learning rate 0.05 and 80% row subsampling
xgb_model, xgb_preds, xgb_rmse, xgb_r2 = evaluate_model(
    'XGBoost',
    XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=6,
                 subsample=0.8, random_state=42, verbosity=0),
    X_train, y_train, X_test, y_test
)
results['XGBoost'] = {'RMSE': xgb_rmse, 'R2': xgb_r2}
```

```python
# ── Plot: Actual vs Predicted — best traditional model (Random Forest) ────────
plt.figure(figsize=(8, 5))
plt.scatter(y_test, rf_preds, alpha=0.3, s=10, color='steelblue', label='Predictions')
max_val = max(y_test.max(), rf_preds.max())
plt.plot([0, max_val], [0, max_val], 'r--', linewidth=1.5, label='Perfect prediction')
plt.title('Random Forest — Actual vs Predicted Weekly Sales')
plt.xlabel('Actual Sales ($)')
plt.ylabel('Predicted Sales ($)')
plt.legend()
plt.tight_layout()
plt.savefig('plot8_rf_actual_vs_predicted.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Section 8 — T3: Deep Learning — MLP (Multi-Layer Perceptron)

We build a feedforward neural network using **PyTorch**.

**Architecture:** 4 layers (256 → 128 → 64 → 1) with ReLU activations, BatchNorm, and Dropout for regularization.  
**Why MLP over LSTM?** Our data is not strictly sequential per store — rows are mixed across stores/departments. An MLP on the tabular feature set is the standard approach here.  
**Important:** Features must be scaled to zero mean / unit variance before training a neural network.

```python
# ── Scale features using StandardScaler ──────────────────────────────────────
# IMPORTANT: fit the scaler ONLY on train data to prevent data leakage
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # fit + transform on train
X_test_scaled  = scaler.transform(X_test)        # transform only on test

# ── Convert numpy arrays to PyTorch tensors ───────────────────────────────────
X_tr_t = torch.tensor(X_train_scaled, dtype=torch.float32)
y_tr_t = torch.tensor(y_train.values, dtype=torch.float32).unsqueeze(1)  # shape: (N, 1)
X_te_t = torch.tensor(X_test_scaled,  dtype=torch.float32)
y_te_t = torch.tensor(y_test.values,  dtype=torch.float32).unsqueeze(1)

# ── Wrap in DataLoader for mini-batch gradient descent ───────────────────────
# Batch size of 512 balances training speed and gradient stability
train_dataset = TensorDataset(X_tr_t, y_tr_t)
train_loader  = DataLoader(train_dataset, batch_size=512, shuffle=True)

print(f'Input features: {X_tr_t.shape[1]}')
print(f'Training batches per epoch: {len(train_loader)}')
```

```python
# ── Define the MLP architecture ───────────────────────────────────────────────
class SalesMLP(nn.Module):
    """
    Feedforward neural network for weekly sales regression.
    Architecture: Input → 256 → 128 → 64 → 1
    Regularization: BatchNorm + Dropout at each hidden layer
    """
    def __init__(self, input_dim):
        super(SalesMLP, self).__init__()
        self.net = nn.Sequential(
            # Layer 1: input → 256 units
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.BatchNorm1d(256),
            nn.Dropout(0.3),       # 30% dropout for regularization

            # Layer 2: 256 → 128 units
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.2),       # 20% dropout

            # Layer 3: 128 → 64 units
            nn.Linear(128, 64),
            nn.ReLU(),

            # Output layer: 64 → 1 (predicted weekly sales)
            nn.Linear(64, 1)
        )

    def forward(self, x):
        return self.net(x)


# ── Instantiate model, loss function, optimizer, and learning rate scheduler ──
input_dim  = X_tr_t.shape[1]
mlp_model  = SalesMLP(input_dim)
criterion  = nn.MSELoss()                                               # MSE loss for regression
optimizer  = torch.optim.Adam(mlp_model.parameters(),                  # Adam optimizer
                               lr=1e-3, weight_decay=1e-4)              # L2 regularization
scheduler  = torch.optim.lr_scheduler.StepLR(optimizer,                # halve LR every 20 epochs
                                              step_size=20, gamma=0.5)

print(mlp_model)
```

```python
# ── Training loop — 60 epochs ─────────────────────────────────────────────────
N_EPOCHS     = 60
train_losses = []  # track average MSE loss per epoch

mlp_model.train()  # set model to training mode (activates Dropout and BatchNorm)

for epoch in range(N_EPOCHS):
    epoch_loss = 0.0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()             # clear gradients from previous step
        preds = mlp_model(X_batch)        # forward pass: compute predictions
        loss  = criterion(preds, y_batch) # compute MSE loss
        loss.backward()                   # backpropagation: compute gradients
        optimizer.step()                  # update model weights
        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_loss)
    scheduler.step()  # decay learning rate

    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch+1:3d}/{N_EPOCHS}]  Avg MSE Loss: {avg_loss:,.0f}')

print('\n Training complete.')
```

```python
# ── Plot: MLP Training Loss Curve ─────────────────────────────────────────────
# A decreasing loss confirms the model is learning
plt.figure(figsize=(8, 4))
plt.plot(range(1, N_EPOCHS + 1), train_losses, color='steelblue')
plt.title('MLP Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Avg MSE Loss')
plt.tight_layout()
plt.savefig('plot10_mlp_loss_curve.png', dpi=150, bbox_inches='tight')
plt.show() 
```

```python
# ── Evaluate MLP on the test set ──────────────────────────────────────────────
mlp_model.eval()  # set to eval mode (disables Dropout and BatchNorm training behavior)

with torch.no_grad():  # disable gradient tracking for inference
    mlp_preds = mlp_model(X_te_t).squeeze().numpy()

mlp_mse  = mean_squared_error(y_test, mlp_preds)
mlp_rmse = np.sqrt(mlp_mse)
mlp_r2   = r2_score(y_test, mlp_preds)

print(f'MLP (Deep Learning) → RMSE: ${mlp_rmse:,.0f}  |  MSE: {mlp_mse:,.0f}  |  R²: {mlp_r2:.4f}')

# Store result for final comparison
results['MLP (Deep Learning)'] = {'RMSE': mlp_rmse, 'R2': mlp_r2}
```
