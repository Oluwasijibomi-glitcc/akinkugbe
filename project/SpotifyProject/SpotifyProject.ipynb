Feature Engineering


```python
data['year'] =  data['track_album_release_date'].str.split('-').str[0]
data['year'] = data['year'].astype(int)
```

```python
data['year'].head()
```

```python
data.isnull().sum()
```

```python
data['duration_minutes'] = data['duration_ms'] / 60000

data[['duration_minutes', 'duration_ms']].head()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.shape
```

```python
# Pattern matches common separators: feat., ft., featuring, with, &, vs., or commas
import re


pattern = r'^(?P<main_artist>[^,(]+?)\s*(?:(?:\(|\[)?\b(?:feat\.?|ft\.?|featuring|with|vs\.?|&)\b|,)\s*(?P<featured_artists>.*)$'

# Extract into two new columns
extracted = data['track_artist'].str.extract(pattern, flags=re.IGNORECASE)

# Clean up trailing closing brackets if they were part of '(feat. ...)'
extracted['featured_artists'] = extracted['featured_artists'].str.rstrip(')]')

# Assign back to the dataframe (falling back to original name if no feature exists)
data['main_artist'] = extracted['main_artist'].fillna(data['track_artist']).str.strip()
data['featured_artists'] = extracted['featured_artists'].str.strip()
```

```python
data['featured_artists'] = data['featured_artists'].fillna('None')
```

```python
data[['track_artist', 'main_artist', 'featured_artists']].head(25)
```

```python
data['featured_artists_count'] = data['featured_artists'].apply(lambda x: len(x.split(',')) if x != 'None' else 0)
data[['track_artist', 'main_artist', 'featured_artists', 'featured_artists_count']].head(25)
```

Audio Feature Interactions

```python
# Energy & Danceability interaction
data['energy_danceability'] = data['energy'] * data['danceability']
# Valence & Danceability (often indicative of overall 'groove' or 'party vibe')
data['valence_danceability'] = data['valence'] * data['danceability']
# Energy & Loudness (overall track intensity)
data['energy_loudness'] = data['energy'] * data['loudness']
# Acousticness & Energy (contrast feature: high acoustic typically opposes high energy)
data['acoustic_energy_ratio'] = data['energy'] / (data['acousticness'] + 1e-5)
```

```python
raw_data = data.copy()
raw_data.drop(columns=dropped_columns, inplace=True)
```

```python
raw_data.info()
```

```python

numeric_features = raw_data.select_dtypes(include=[np.number]).columns.tolist()
calculated_features = ['duration_minutes', 'featured_artists_count', 'energy_danceability', 'valence_danceability', 'energy_loudness', 'acoustic_energy_ratio']
categorical_features = raw_data.select_dtypes(include=[object]).columns.tolist()
y = raw_data['is_popular']
X = raw_data.drop(columns=['is_popular'])
```

```python

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.neural_network import MLPClassifier
from sklearn import linear_model

```

 Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42, stratify=y)    

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

Logistic Regression

```python
clf = LogisticRegression(C=1e5, solver='lbfgs', max_iter=1000)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Random Forest

```python
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
rf_y_pred = rf.predict(X_test)

print("Random Forest Accuracy:", accuracy_score(y_test, rf_y_pred))
print(classification_report(y_test, rf_y_pred))
```

XGBoost

```python
xgb = XGBClassifier()
xgb.fit(X_train, y_train)
xgb_y_pred = xgb.predict(X_test)    

print("XGBoost Accuracy:", accuracy_score(y_test, xgb_y_pred))
print(classification_report(y_test, xgb_y_pred))
```

Neural Network

```python
neural_net = MLPClassifier(hidden_layer_sizes=(100,), max_iter=500, random_state=42)
neural_net.fit(X_train, y_train)
nn_y_pred = neural_net.predict(X_test)  

print("Neural Network Accuracy:", accuracy_score(y_test, nn_y_pred))
print(classification_report(y_test, nn_y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay

# 1. Create a single plot layout
fig, ax = plt.subplots(figsize=(7, 7))

# 2. Plot each previously trained model onto the same 'ax'
RocCurveDisplay.from_estimator(rf, X_test, y_test, ax=ax, name="Random Forest")
RocCurveDisplay.from_estimator(clf, X_test, y_test, ax=ax, name="Logistic Regression")
RocCurveDisplay.from_estimator(xgb, X_test, y_test, ax=ax, name="XGBoost")
RocCurveDisplay.from_estimator(neural_net, X_test, y_test, ax=ax, name="Neural Network")

# 3. Add a baseline reference line (random guessing)
ax.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Baseline (AUC = 0.50)")

# 4. Final plot customizations
ax.set_title("ROC Curve Comparison")
ax.legend(loc="lower right")
plt.grid(True)
plt.show()
```

SHAP Importance 

```python
import shap
```

```python
background_data = shap.utils.sample(X_train, 100, random_state=42)
```

```python

explainer_xgb = shap.TreeExplainer(xgb)
shap_values_xgb = explainer_xgb(X_test)
if len(shap_values_xgb.shape) == 3: shap_values_xgb = shap_values_xgb[:, :, 1]



explainer_rf = shap.TreeExplainer(rf)
shap_values_rf = explainer_rf(X_test)
if len(shap_values_rf.shape) == 3: shap_values_rf = shap_values_rf[:, :, 1]


 
explainer_lr = shap.LinearExplainer(clf, background_data)
shap_values_lr = explainer_lr(X_test)


nn_sample_test = X_test.sample(50, random_state=42)
explainer_nn = shap.KernelExplainer(neural_net.predict_proba, background_data)
shap_values_nn = explainer_nn.shap_values(nn_sample_test)
shap_values_nn_class1 = shap_values_nn[1] if isinstance(shap_values_nn, list) else shap_values_nn[:, :, 1]


shap_values_nn_obj = shap.Explanation(
    values=shap_values_nn_class1,
    base_values=explainer_nn.expected_value[1],
    data=nn_sample_test.values,
    feature_names=X_test.columns
)

print("SHAP calculations complete!")


fig, axes = plt.subplots(2, 2, figsize=(16, 16))


plt.axes(axes[0, 0])
plt.title("XGBoost Global Importance", fontsize=14, pad=10)
shap.plots.bar(shap_values_xgb, show=False)


plt.axes(axes[0, 1])
plt.title("Random Forest Global Importance", fontsize=14, pad=10)
shap.plots.bar(shap_values_rf, show=False)


plt.axes(axes[1, 0])
plt.title("Logistic Regression Global Importance", fontsize=14, pad=10)
shap.plots.bar(shap_values_lr, show=False)


plt.axes(axes[1, 1])
plt.title("Neural Network Global Importance", fontsize=14, pad=10)
shap.plots.bar(shap_values_nn_obj, show=False)

plt.tight_layout()
plt.show()
```
